```python
# Load Model

with open('num_cols.txt', 'r') as file_1:
  num_cols = json.load(file_1)

with open('cat_cols.txt', 'r') as file_2:
  cat_cols = json.load(file_2)

with open('best_rf_model.pkl', 'rb') as file_3:
  best_rf_model = pickle.load(file_3)
```

```python
# Input data
data = {
    'CreditScore': [700, 820],
    'Geography': ["Germany", "Germany"],
    'Gender': ["Female", "Male"],
    'Age': [33, 29],
    'Tenure': [7, 5],
    'Balance': [130000, 97000],
    'NumOfProducts': [3, 1],
    'HasCrCard': [1, 0],
    'IsActiveMember': [1, 1],
    'EstimatedSalary': [85000, 90000]
}

df = pd.DataFrame(data)
df
```

```python
# Splitt kolom
data_inf_num = df[num_cols]
data_inf_cat = df[cat_cols]
data_inf_num
```

```python
#Gabungkan kembali kolom numerik dan kategorikal
data_final = pd.concat([data_inf_num, data_inf_cat], axis=1)
```

```python
#Lakukan prediksi menggunakan model yang telah dimuat
predictions = best_rf_model.predict(data_final)

# Looping untuk memberikan gambaran jelas
churn_labels = []
for pred in predictions:
    if pred == 1:
        churn_labels.append('Churn')
    else:
        churn_labels.append('Tidak Churn')

#Tampilkan hasil prediksi
print(f"Hasil prediksi untuk data pertama : ", churn_labels[0])
print(f"Hasil prediksi untuk data kedua : ", churn_labels[1])
```
